```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        rdist = math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)
        return rdist
```

```python
class Rectangle:
    def __init__(self):
        self.c1 = Point(0, 0)
        self.c2 = Point(0, 2)
        self.c3 = Point(2, 0)
        self.c4 = Point(2, 2)
        self.width = 0
        self.height = 0

    def calculate_width(self):
        self.width = self.c1.distance(self.c3)
        return self.width

    def calculate_height(self):
        self.height = self.c1.distance(self.c2)
        return self.height

    def area(self):
        self.calculate_width()
        self.calculate_height()
        return self.width * self.height

square = Rectangle()
print(square.area())
```
