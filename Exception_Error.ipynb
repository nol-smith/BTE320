```python
# PARKING GARAGE FEES

print("Parking Garage Fees")

def calcFee(hours):
    fee = 5 + 2.5 * hours
    if fee < 10:
        fee = 10
    elif fee > 20:
        fee = 20
    print(f'Fee: {fee}$')

def main():
    try:
        numHours = int(input("Number of hours parked: "))
        calcFee(numHours)
    except ValueError:
        print("Invalid input. Please enter a valid number of hours.")
    except TypeError:
        print("Invalid input. Please enter a valid number of hours.")

if __name__ == "__main__":
    main()
```

```python
def getValues(n):
    L = []
    for i in range (n):
      weight = float(input("Enter weight: "))
      L.append(weight)
    return L

def calcAverage(alist):
  print(f'The average weight is {sum(alist[2:])/len(alist[2:])}')

if __name__ == "__main__":
  try:
    n = int(input("Enter number of weights: "))
    alist = getValues(n)
    calcAverage(alist)
  except ValueError:
    print("Invalid input. Please enter a valid number of weights.")
  except TypeError:
    print("Invalid input. Please enter a valid number of weights.")
  except ZeroDivisionError:
    print("Invalid input. Please enter a valid number of weights.")
```
