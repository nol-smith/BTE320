```python
# PARKING GARAGE FEES
print("Parking Garage Fees")
hours = int(input("How many hours did you park? "))

def parking_fees(hours):
    if hours >= 6:
        print(20)
    elif hours >= 3:
        print((hours * 2.50) + 5)
    elif hours <= 2:
        print(10)

parking_fees(hours)

```
