Reverse Fibonacci

```python
def rfib(n, a=0, b=1):
    if n == 0:
        return []
    return [b] + rfib(n - 1, b, a + b)

n = 10  # Number of terms
print(rfib(n))
```

Reverse a String

```python
def reverse(s):
  if len(s) == 1:
    return s
  else:
    return s[-1] + reverse(s[:-1])

reverse('Hello')
```
