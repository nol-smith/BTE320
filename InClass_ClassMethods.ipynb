```python
class Greeting:
  def __init__(self, first, last, formal):
    self.first = first
    self.last = last
    self.formal = formal

  def display(self):
    if self.formal:
      print(f'Hello {self.first} {self.last}')
    else:
      print(f'Hi {self.first}')

greeting = Greeting('Donald','Becker', formal=False)
greeting.display()
```

if formal is true, display hello <first><last>
